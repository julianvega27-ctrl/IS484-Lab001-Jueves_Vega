# EXTENSIÓN PARA LA CASA

## PARTE A: FICHA PEAS
> **Objetivo:** Mantener la humedad del suelo en un rango saludable, sin desperdiciar agua.
> 
> **Medida de Desempeño (P):** Tiempo que la humedad se mantiene en un rango óptimo, agua utilizada, riesgo duplicado (regar y que llueva).
> 
> **Entorno (E):** Jardin exterior con clima variable y predecible.
> 
> **Acciones (A):** Regar o no regar
> 
> **Percepciòn (S):** humedad del suelo (humedad_suelo) y si hoy lloverá (llovera_hoy)

## PARTE B: REGLAS DE DESICIÒN
> **Condición:** huemadad_suelo<30 Y no llovera_hoy **Acción:** Regar.
> 
>  **Condición:** huemadad_suelo<30 y llovera_hoy **Acción:** No regar, se espera que llueva.
> 
>  **Condición:** humedad_suelo>=30 **Acción:** No regar.

## PARTE C: CÓDIGO

```python
def agente_riego (humedad_suelo, llovera_hoy):
    if humedad_suelo < 30 and not llovera_hoy:
        return "Regar"
    elif humedad_suelo < 30 and llovera_hoy:
        return "No regar, se espera que llueva."
    else:
        return "No regar."
```

```python
casos = [(15, False),(15, True),(60, False),(29, False),(30, False),]

for humedad_suelo, llovera_hoy in casos:
    accion = agente_riego(humedad_suelo, llovera_hoy)
    print(f"Humedad: {humedad_suelo:>3} | Llovera hoy: {llovera_hoy!s:>5} -> Accion: {accion}")
```
