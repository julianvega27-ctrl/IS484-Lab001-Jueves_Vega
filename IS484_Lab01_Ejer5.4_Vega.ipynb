```python
def agente_termostato (temperatura_actual, temperatura_objetivo=22, margen=1.0):
    """
    Agente reactivo simple (sin memoria, sin modelo del entorno).

    Analisis PEAS (Seccion 5.
    - Objetivo: mantener la temperatura cerca de temperatura_objetivo (+/- margen).

    - Medida de desempeño (P): minimizar la diferencia entre temperatura_actual y
      temperatura_objetivo a lo largo del tiempo, evitando encendidos innecesarios
      (oscilaciones) que desperdicien energia.
      
    - Entorno (E): habitacion cerrada cuya temperatura varia con el tiempo y el clima
      exterior; parcialmente observable, ya que el agente solo percibe la lectura
      del sensor y no la causa del cambio de temperatura.2nte.
    - Acciones (A): encender aire acondicionado, encender calefaccion, no hacer ne    - Percepcion (S): temperatura_actual medida por el sensor del ambiente. ien energia.
    """  
    diferencia = temperatura_actual - temperatura_objetivo

    if diferencia > margen:
        return "Encender aire acondicionado"
    elif diferencia < -margen:
        return "Encender calefacción"
    else:
        return "No hacer nada"
```

```python
percepciones = [18.0, 19.5, 21.8, 24.3, 26.0, 23.0, 20.5, 21.0, 27.5]

for t in percepciones:
    accion = agente_termostato(t)
    print(f"Percepcion: {t} C -> Accion del agente: {accion}")
```

1)	¿Este agente termostato es racional?
  
  Sí, ya que dado el entorno para el que fue diseñado donde su única percepción es la temperatura actual, siempre debe elegir la acción que reduce la diferencia respecto a la temperatura objetivo. Esto es lo que exige la medida de desempeño.
  
2)¿Bajo qué condiciones del entorno dejaría de serlo (por ejemplo, si el sensor de temperatura falla o si el margen es demasiado amplio)?

•	Si el sensor de temperatura falla o s entrega lecturas erróneas, el agente seguirá actuando “correctamente” según su regla, pero sobre una percepción falsa, por lo que sus acciones dejarán de ser racionales respecto al estado real del ambiente.

•	Si el margen es demasiado amplio, el agente tolerará diferencias de temperatura que ya resultan incómodas para las personas, dejando de maximizar realmente su medida de desempeño.

•	Si el ambiente del entorno cambia muy rápido y el agente no tiene memoria ni capacidad de anticipar cambios, puede reaccionar tarde, encendiendo y apagando el sistema de forma ineficiente.
